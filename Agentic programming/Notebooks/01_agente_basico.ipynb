### Agente ReAct (Reasoning and Acting) Conceptual

Un agente ReAct alterna entre dos fases:
1.  **Razonamiento (Reasoning):** El agente piensa qué hacer a continuación, basándose en la observación actual y su historial de interacciones.
2.  **Actuación (Acting):** El agente ejecuta una acción, que puede ser usar una herramienta, o emitir una respuesta final.

```python
import time

# --- Definición de Herramientas (Tools) ---
# Las herramientas son funciones que el agente puede 'llamar' para interactuar con el mundo.

def get_current_time(query=None):
    """Devuelve la hora actual del sistema."""
    return f"La hora actual es: {time.strftime('%H:%M:%S')}"

def search_knowledge_base(query):
    """Simula una búsqueda en una base de conocimientos. """
    query = query.lower()
    if "python" in query:
        return "Python es un lenguaje de programación de alto nivel, interpretado y de propósito general."
    elif "react" in query:
        return "React es una biblioteca de JavaScript para construir interfaces de usuario."
    else:
        return f"No se encontró información sobre '{query}'."

# Diccionario de herramientas disponibles para el agente
tools = {
    "get_current_time": get_current_time,
    "search_knowledge_base": search_knowledge_base,
}

# --- Clase Agente ReAct ---
class ReActAgent:
    def __init__(self, tools):
        self.tools = tools
        self.history = []

    def _reason(self, observation):
        """Simula la fase de razonamiento. En un agente real, esto sería manejado por un LLM."""
        print(f"\nObservación: {observation}")
        print("--- Pensando ---")

        # Ejemplo de lógica de razonamiento simple (un LLM haría esto de forma más inteligente)
        if "hora" in observation.lower():
            return "Necesito saber la hora actual. Usaré la herramienta 'get_current_time'."
        elif "qué es" in observation.lower() or "información sobre" in observation.lower():
            keyword = observation.lower().split("qué es ")[-1].split(" información sobre ")[-1].replace('?','')
            return f"Necesito buscar información sobre '{keyword}'. Usaré la herramienta 'search_knowledge_base' con el argumento '{keyword}'."
        elif "gracias" in observation.lower():
            return "El usuario está agradeciendo. Es hora de terminar con un mensaje de despedida."
        else:
            return "No estoy seguro de cómo responder. Pediré más información o daré una respuesta genérica."

    def _act(self, thought):
        """Simula la fase de actuación. Un LLM generaría el comando de la herramienta."""
        print(f"Pensamiento: {thought}")
        print("--- Actuando ---")

        # Parsear el pensamiento para decidir la acción
        if "usaré la herramienta 'get_current_time'" in thought.lower():
            tool_name = "get_current_time"
            result = self.tools[tool_name]()
            return {"type": "tool_output", "tool": tool_name, "output": result}
        elif "usaré la herramienta 'search_knowledge_base'" in thought.lower():
            tool_name = "search_knowledge_base"
            # Extraer el argumento (esto sería más robusto con un LLM)
            try:
                arg = thought.split("con el argumento '")[1].split("'" )[0]
                result = self.tools[tool_name](arg)
                return {"type": "tool_output", "tool": tool_name, "output": result}
            except IndexError:
                return {"type": "response", "output": "Error al extraer el argumento para la búsqueda."}
        elif "es hora de terminar" in thought.lower():
            return {"type": "response", "output": "¡De nada! Estoy aquí para ayudarte."}
        else:
            return {"type": "response", "output": "Lo siento, no pude entender tu solicitud. ¿Puedes reformularla?"}

    def run(self, initial_observation, max_steps=5):
        observation = initial_observation
        for step in range(max_steps):
            print(f"\n--- Paso {step + 1} ---")

            # Razonar
            thought = self._reason(observation)
            self.history.append(f"Pensamiento: {thought}")

            # Actuar
            action = self._act(thought)
            self.history.append(f"Acción: {action}")

            if action["type"] == "tool_output":
                print(f"Resultado de la herramienta '{action['tool']}': {action['output']}")
                observation = f"Resultado de la herramienta '{action['tool']}': {action['output']}"
            elif action["type"] == "response":
                print(f"Agente: {action['output']}")
                return action['output'] # El agente ha respondido, terminamos
            else:
                print("Acción desconocida.")
                return "Error interno del agente."

        return "Se alcanzó el número máximo de pasos sin una respuesta final."

print("Agente ReAct inicializado. Puedes crear una instancia y ejecutarlo.")
```

### Ejemplo de uso del Agente ReAct

Aquí puedes interactuar con el agente. Observa cómo el agente 'piensa' y luego 'actúa' (ya sea usando una herramienta o dando una respuesta).

```python
# Crea una instancia del agente
agent = ReActAgent(tools)

# --- Prueba 1: Preguntar la hora ---
print("\n--- Escenario 1: Preguntar la hora ---")
agent.run("¿Cuál es la hora actual?")

# --- Prueba 2: Preguntar sobre un tema ---
print("\n--- Escenario 2: Preguntar sobre un tema ---")
agent.run("¿Qué es Python?")

# --- Prueba 3: Preguntar sobre otro tema ---
print("\n--- Escenario 3: Preguntar sobre otro tema ---")
agent.run("Necesito información sobre React")

# --- Prueba 4: Pregunta que no sabe responder ---
print("\n--- Escenario 4: Pregunta genérica ---")
agent.run("Cuéntame un cuento.")

# --- Prueba 5: Despedida ---
print("\n--- Escenario 5: Despedida ---")
agent.run("Gracias por tu ayuda.")
```

### Notas importantes:

*   **Simplificación:** En un agente ReAct real, la fase de `_reason` y la lógica para parsear el `_act` serían gestionadas por un Modelo de Lenguaje Grande (LLM) que generaría texto siguiendo un formato específico (por ejemplo, `Thought: ... Action: tool_name[tool_argument]`).
*   **Herramientas:** Las `tools` pueden ser cualquier función Python que el LLM pueda entender y llamar, como APIs externas, bases de datos, etc.
*   **Loop:** El agente ReAct funciona en un bucle: Observación -> Razonamiento -> Acción -> Nueva Observación (resultado de la acción) y repite hasta que se genera una respuesta final o se agotan los pasos.
